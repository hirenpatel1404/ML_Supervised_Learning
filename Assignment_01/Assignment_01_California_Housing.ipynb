# Assignment 1: Downloading, Loading, and Exploring the California Housing Dataset

In this assignment, I will load the California Housing dataset from scikit-learn and explore its structure. This dataset comes from the 1990 U.S. Census and is used to predict the median house value in California districts based on socio-economic and geographical factors.

My steps:
- Load the dataset and understand what the features and target represent
- Check the shape of the dataset
- Display basic statistics
- Check for missing values

## Step 1: Load the California Housing Dataset

```python
# Import libraries
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
# Load the California Housing dataset as a pandas DataFrame
housing = fetch_california_housing(as_frame=True)
```

```python
# Store the full table (features + target) in df
df = housing.frame
```

```python
# Show the feature names and target name
print("Features:", housing.feature_names)
print("Target:", housing.target_names)
```

```python
# Show the first 5 rows
df.head()
```

### Interpretation

I loaded the California Housing dataset using fetch_california_housing(as_frame=True), which gave me the data as a pandas DataFrame. Each row represents one district (a census block group) in California, not a single house.

What the features represent (the inputs the model will use):
- MedInc: median income of households in the district, in tens of thousands of dollars (8.3252 means $83,252)
- HouseAge: median age of the houses in the district, in years
- AveRooms: average number of rooms per household
- AveBedrms: average number of bedrooms per household
- Population: total number of people living in the district
- AveOccup: average number of people living in each household
- Latitude: how far north the district is
- Longitude: how far west the district is

AveRooms, AveBedrms, and AveOccup have decimal values because they are averages across all households in the district. For example, if three homes have 2, 3, and 3 bedrooms, the average is 2.67.

What the target represents (what I want to predict):
- MedHouseVal: the median house value in the district, in hundreds of thousands of dollars (4.526 means $452,600)

Since the target is a continuous number, this is a regression problem.

## Step 2: Check the Shape of the Dataset

```python
# Check the number of rows and columns
print("Shape of the dataset:", df.shape)
print("Number of rows (districts):", df.shape[0])
print("Number of columns:", df.shape[1])

# Check column data types and non-null counts
df.info()
```

### Interpretation

The dataset has 20,640 rows and 9 columns. This means I have data for 20,640 districts in California, with 8 feature columns and 1 target column (MedHouseVal).

From df.info():
- All 9 columns are float64, which means they are all decimal numbers. There are no text or categorical columns, so I will not need encoding (like one-hot encoding) before modelling.
- Every column shows 20,640 non-null values, which matches the total number of rows. This is a first sign that there are no missing values. I will confirm this properly in Step 4.
- The dataset uses about 1.4 MB of memory, so it is small and fast to work with on my laptop.

## Step 3: Display Basic Statistics

```python
# Show summary statistics for every column
df.describe().round(2)
```

### Interpretation

df.describe() gave me the count, mean, standard deviation, minimum, maximum, and percentiles for each column.

Key observations:
- Count is 20,640 for every column, which again suggests there are no missing values.
- MedInc (median income) ranges from 0.50 to 15.00, with a median of 3.53 ($35,300). The mean (3.87) is higher than the median, which means a few high-income districts pull the average up (right skew).
- MedHouseVal (the target) ranges from 0.15 ($15,000) to 5.00 ($500,000), with a median of 1.80 ($180,000). Again the mean (2.07) is higher than the median, so house values are right-skewed.
- Some columns appear to be capped. HouseAge stops at exactly 52, MedInc stops at exactly 15.00, and MedHouseVal stops at exactly 5.00. This suggests that values above these limits were recorded as the maximum. For the target, this means the model cannot learn about houses worth more than $500,000.
- There are extreme outliers in AveRooms (max 141.91 compared to a median of 5.23), AveBedrms (max 34.07 compared to a median of 1.05), and AveOccup (max 1,243 people per household compared to a median of 2.82). These are likely districts with very few households but large numbers of people or rooms, such as dorms or institutions.
- Population also varies a lot, from 3 to 35,682 people, with a median of 1,166.
- Latitude (32.54 to 41.95) and Longitude (-124.35 to -114.31) cover the full area of California, from south to north and from the coast to the eastern border.

These capped values and outliers could affect model performance later, so they may need to be handled during preprocessing.

## Step 4: Check for Missing Values

```python
# Count missing values in each column
print("Missing values per column:")
print(df.isnull().sum())

# Total missing values in the whole dataset
print("\nTotal missing values:", df.isnull().sum().sum())
```

### Interpretation

I used df.isnull().sum() to count missing values in each column, and df.isnull().sum().sum() to get the total for the whole dataset.

Every column shows 0 missing values, and the total is 0. This confirms what I saw in df.info() and df.describe(), where every column had 20,640 values. Because of this, I do not need to remove rows or fill in missing values (imputation) before building a model.
